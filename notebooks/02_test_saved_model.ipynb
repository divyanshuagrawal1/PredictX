```python
import joblib
```

```python
loaded_model = joblib.load("../models/xgboost_model.pkl")

print("Model loaded successfully!")
```

```python
loaded_preprocessor = joblib.load("../models/preprocessor.pkl")

print("Preprocessor loaded successfully!")
```

```python
import json

with open("../models/config.json", "r") as file:
    config = json.load(file)

threshold = config["threshold"]

print("Threshold:", threshold)
```

```python
import pandas as pd
new_machine = pd.DataFrame({
    "Type": ["L"],
    "Air temperature [K]": [300.0],
    "Process temperature [K]": [310.0],
    "Rotational speed [rpm]": [1500],
    "Torque [Nm]": [40.0],
    "Tool wear [min]": [100]
})

new_machine
```

```python
new_machine_processed = loaded_preprocessor.transform(new_machine)

print("New machine processed successfully!")
print(new_machine_processed)
```

```python
failure_probability = loaded_model.predict_proba(new_machine_processed)[0][1]

print("Failure probability:", failure_probability)
```

```python
prediction = int(failure_probability >= threshold)

print("Prediction:", prediction)
```

```python
if prediction == 1:
    print("⚠️ Machine Failure Predicted")
else:
    print("✅ Machine is Normal")
```
